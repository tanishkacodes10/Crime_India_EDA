```python
df['Date Reported'] = pd.to_datetime(
    df['Date Reported'],
    dayfirst=True,
    errors='coerce'
)

df['Date of Occurrence'] = pd.to_datetime(
    df['Date of Occurrence'],
    dayfirst=True,
    errors='coerce'
)
```

```python
df['Year'] = df['Date Reported'].dt.year
df['Month'] = df['Date Reported'].dt.month
df['Month Name'] = df['Date Reported'].dt.month_name()
```

```python
df['Year'].value_counts().sort_index()
```

```python
monthly_crimes = df['Month'].value_counts().sort_index()

plt.figure(figsize=(10,6))
monthly_crimes.plot(kind='bar')
plt.title("Crime Cases by Month")
plt.xlabel("Month")
plt.ylabel("Number of Cases")
plt.show()
```

```python
df['Time of Occurrence'] = pd.to_datetime(
    df['Time of Occurrence'],
    format='%H:%M',
    errors='coerce'
)

df['Hour'] = df['Time of Occurrence'].dt.hour
```

```python
df['Hour'].value_counts().sort_index()
```

```python
pd.crosstab(
    df['Crime Domain'],
    df['Victim Gender']
)
```

```python
pd.crosstab(
    df['Crime Domain'],
    df['Case Closed']
)
```

```python
top_cities = df['City'].value_counts().head(10).index

city_crime = pd.crosstab(
    df[df['City'].isin(top_cities)]['City'],
    df[df['City'].isin(top_cities)]['Crime Domain']
)

city_crime
```
